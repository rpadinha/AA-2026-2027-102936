**P25 - Min Weight Dominating Set:** 
Find a minimum weight dominating set for a given
undirected graph G(V, E), whose vertices carry positive weights, with n vertices and m edges. A
dominating set of G is a subset D of vertices, such that every vertex not in D is adjacent to, at least,
one vertex in D. The weight of a dominating set is the sum of its vertices’ weights. A minimum
weight dominating set is a dominating set whose total weight is as small as possible.
